```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000

data = {
    "Transaction_ID": range(1, n + 1),
    "Customer_ID": np.random.randint(1001, 1151, n),
    "Transaction_Date": pd.date_range(
        start="2025-01-01",
        periods=n,
        freq="h"
    ),
    "Category": np.random.choice(
        ["Groceries", "Shopping", "Transport",
         "Entertainment", "Utilities"],
        n
    ),
    "Amount": np.random.uniform(50, 10000, n).round(2),
    "Payment_Method": np.random.choice(
        ["Card", "EFT", "Mobile Payment"],
        n
    )
}

df = pd.DataFrame(data)

df.head()
```
